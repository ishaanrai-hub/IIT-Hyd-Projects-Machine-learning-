Question 15 – k-NN Implementation with Optimization (Theory Only)
Part A: Feature Scaling (Main Theory Points)
🔹 Why Feature Scaling is Essential for k-NN

k-NN is a distance-based algorithm that relies on metrics such as Euclidean distance.

In the Breast Cancer dataset, features like area_mean and perimeter_mean have much larger numerical ranges than features like smoothness_mean.

Without feature scaling, larger-scale features dominate distance calculations, leading to incorrect nearest neighbors.

Standardization ensures all features contribute equally to the distance.

🔹 Standardization (Mean = 0, Std = 1)

Standardization rescales each feature to have:

Mean = 0

Standard deviation = 1

This improves:

Model performance

Fair distance comparison

Convergence stability

🔹 Impact of Scaling

k-NN accuracy is significantly lower without scaling

After scaling, accuracy improves noticeably, confirming that feature scaling is essential for k-NN

Part B: Finding Optimal k (Main Theory Points)
🔹 Systematic Approach for Selecting k

Odd values of k from 1 to 50 are tested to avoid ties in classification

5-fold cross-validation is used to ensure reliable and unbiased performance evaluation

Validation accuracy is averaged across folds for each k

🔹 Accuracy vs k Analysis

Small k values (e.g., k=1) show:

Very high training accuracy

Low validation accuracy

High variance (overfitting)

Large k values show:

Lower training and validation accuracy

High bias (underfitting)

🔹 Training vs Validation Accuracy

The gap between training and validation accuracy decreases as k increases

The optimal k is chosen where validation accuracy is highest and the gap is minimal

🔹 Bias–Variance Tradeoff

Low k → low bias, high variance

High k → high bias, low variance

Optimal k provides a balance between bias and variance, resulting in best generalization

Final Conclusion (Easy to Remember)

Feature scaling is critical for k-NN due to its reliance on distance metrics. Using cross-validation to select an optimal k helps balance the bias–variance tradeoff and significantly improves classification performance on the Breast Cancer dataset.

Question 16: SVM and Decision Tree with Class Imbalance (Theory)
Dataset Imbalance Setup (Context)

Original dataset is moderately imbalanced.

Modified to 90% benign, 10% malignant using undersampling.

This simulates a real-world medical scenario where malignant cases are rare.

📌 Key Risk:
Accuracy becomes misleading because a model can predict “benign” always and still get ~90% accuracy.

Part A: Handling Class Imbalance (Main Points)
🔹 Why Class Imbalance Is a Problem

Standard classifiers are biased toward the majority class (benign).

Minority class (malignant) is more important but often misclassified.

Leads to high false negatives, which are dangerous in medical diagnosis.

🔹 Model Configurations Used
1. Baseline (No Class Weighting)

Treats both classes equally.

Tends to:

Favor benign class

Miss malignant cases

High accuracy but poor recall for malignant tumors

2. Balanced Class Weights

Automatically adjusts weights inversely proportional to class frequencies.

Forces the model to pay more attention to malignant cases.

Improves:

Recall

F1-score

Slight drop in accuracy (acceptable in healthcare)

3. Custom Class Weights

Weights assigned based on imbalance ratio (e.g., malignant > benign).

Gives maximum importance to detecting cancer.

Reduces false negatives further.

May increase false positives (acceptable tradeoff in medicine).

🔹 SVM vs Decision Tree (Imbalance Handling)
Model	Strength
SVM	Strong margin control, better generalization
Decision Tree	Interpretable, but more sensitive to imbalance
Part B: Comprehensive Evaluation Strategy (Main Points)
🔹 Why Accuracy Is NOT Enough

With 90:10 imbalance:

90% accuracy can mean zero cancer detected

Hence, accuracy is misleading

🔹 Important Metrics (Medical Priority Order)

Recall (Sensitivity)

Most important metric

Measures how many actual malignant cases are correctly detected

Low recall = missed cancer (very dangerous)

Precision

Measures correctness of malignant predictions

Important, but secondary to recall

F1-Score

Balances precision and recall

Useful under class imbalance

ROC-AUC

Measures discrimination ability across thresholds

Threshold-independent

Average Precision Score

Better than ROC-AUC for imbalanced datasets

Focuses on minority class performance

Visualization Insights (Theory)
🔹 Confusion Matrix

Balanced and custom-weighted models show:

Fewer false negatives

More true positives for malignant class

🔹 ROC Curves

Models with class weighting dominate baseline curves

Higher AUC indicates better separation

🔹 Precision–Recall Curves

More informative than ROC under imbalance

Weighted models show higher area under PR curve

🔹 Bar Charts

Show recall and F1-score improvements clearly

Baseline models perform worst on minority class metrics

Best Model Recommendation

✅ Best Choice:

SVM with balanced or custom class weights

Why?

Highest recall for malignant cases

Strong ROC-AUC and Average Precision

Better generalization than Decision Trees

Fewer missed cancer cases

Optimal Classification Threshold (Important Point)

Default threshold = 0.5

In medical diagnosis:

Lower threshold (e.g., 0.3–0.4) is preferred

This:

Increases recall

Reduces false negatives

Accepts more false positives (safer)

📌 Medical Rule:

It is better to raise a false alarm than to miss cancer.

Final Exam-Ready Conclusion

In highly imbalanced medical datasets, accuracy is unreliable. Models must be evaluated using recall, F1-score, ROC-AUC, and precision-recall analysis. Class weighting significantly improves minority class detection. Among all configurations, SVM with balanced or custom class weights provides the best tradeoff between sensitivity and generalization, making it most suitable for breast cancer diagnosis.
