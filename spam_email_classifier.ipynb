```python
##importing libraries
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score

from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import LinearSVC
from sklearn.tree import DecisionTreeClassifier
```

```python
# 1. Create Synthetic Dataset
# -----------------------------

np.random.seed(42)

n_samples = 5000
n_spam = 500
n_ham = 4500

# Features:
# - 20 word frequency features
# - 1 URL presence feature
# - 1 email length feature
```

```python
# Non-spam emails
ham_word_freq = np.random.normal(loc=1.0, scale=0.5, size=(n_ham, 20))
ham_urls = np.random.binomial(1, 0.1, size=(n_ham, 1))
ham_length = np.random.normal(loc=300, scale=50, size=(n_ham, 1))
```

```python
# Spam emails
spam_word_freq = np.random.normal(loc=2.5, scale=0.7, size=(n_spam, 20))
spam_urls = np.random.binomial(1, 0.8, size=(n_spam, 1))
spam_length = np.random.normal(loc=600, scale=100, size=(n_spam, 1))
```

```python
# Combine features
X_ham = np.hstack([ham_word_freq, ham_urls, ham_length])
X_spam = np.hstack([spam_word_freq, spam_urls, spam_length])

X = np.vstack([X_ham, X_spam])
y = np.array([0] * n_ham + [1] * n_spam)  # 0 = non-spam, 1 = spam
```

```python
# 2. Train-Test Split
# -----------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
```

```python
# 3. k-NN Model
# -----------------------------

knn_model = Pipeline([
    ("scaler", StandardScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=5))
])

knn_model.fit(X_train, y_train)
knn_pred = knn_model.predict(X_test)

print("k-NN Results")
print("Accuracy:", accuracy_score(y_test, knn_pred))
print(classification_report(y_test, knn_pred))
```

```python
# 4. SVM Model
# -----------------------------

svm_model = Pipeline([
    ("scaler", StandardScaler()),
    ("svm", LinearSVC(class_weight="balanced", random_state=42))
])

svm_model.fit(X_train, y_train)
svm_pred = svm_model.predict(X_test)

print("SVM Results")
print("Accuracy:", accuracy_score(y_test, svm_pred))
print(classification_report(y_test, svm_pred))
```

```python
# 5. Decision Tree Model
# -----------------------------

dt_model = DecisionTreeClassifier(
    max_depth=10,
    class_weight="balanced",
    random_state=42
)

dt_model.fit(X_train, y_train)
dt_pred = dt_model.predict(X_test)

print("Decision Tree Results")
print("Accuracy:", accuracy_score(y_test, dt_pred))
print(classification_report(y_test, dt_pred))
```

```python
# 3. Expected Outcome (Typical)
# Model	Accuracy /Spam Recall	 / Notes
# k-NN	Medium	Lower	Struggles with high-dimensional data
# SVM	High	High	Best overall performance
# Decision Tree	Medium–High	Medium	Interpretable but overfits
```

```python
# ✅ Linear SVM is the best choice for spam email classification due to its strong performance with high-dimensional text features and robustness to class imbalance.
# Decision Trees are useful for explainability, while k-NN is best kept as a baseline model.
```
